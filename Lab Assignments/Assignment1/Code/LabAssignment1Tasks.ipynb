```python
import numpy as np
import matplotlib.pyplot as plt

print("Part A")
print("Perceptron Class")
class Perceptron:
    def __init__(self):
        self.weights = None
        self.bias = 0
        self.error_history = []
    def fit(self,X,y,epochs,lr):
        self.weights = np.zeros(X.shape[1])
        self.bias = 0
        self.error_history = []
        for epoch in range(epochs):
            errors = 0
            for i in range(len(X)):
                prediction = self.predict(X[i])
                error = y[i] - prediction
                self.weights += lr * error * X[i]
                self.bias += lr * error
                if error != 0:
                    errors += 1
            self.error_history.append(errors)
            print("Epoch:",epoch + 1)
            print("Misclassifications:",errors)
    def predict(self,X):
        linear_output = np.dot(X,self.weights) + self.bias
        prediction = 1 if linear_output >= 0 else 0
        return prediction
print("Perceptron class created")

print("\nPart B")
print("AND Truth Table")
X_and = np.array([[0,0],[0,1],[1,0],[1,1]])
y_and = np.array([0,0,0,1])
and_perceptron = Perceptron()
epochs = 20
learning_rate = 0.1
print("Training AND perceptron")
and_perceptron.fit(X_and,y_and,epochs,learning_rate)
and_predictions = np.array([and_perceptron.predict(x) for x in X_and])
print("\nAND Results")
print("Input1 Input2 Actual Predicted")
for i in range(len(X_and)):
    print(X_and[i,0],X_and[i,1],y_and[i],and_predictions[i])
print("Final weights:",and_perceptron.weights)
print("Final bias:",and_perceptron.bias)

print("\nPart C")
print("AND Decision Boundary")
x_values = np.linspace(-0.5,1.5,100)
if and_perceptron.weights[1] != 0:
    y_values = -(and_perceptron.weights[0] * x_values + and_perceptron.bias) / and_perceptron.weights[1]
plt.figure(figsize=(7,5))
plt.scatter(X_and[y_and == 0,0],X_and[y_and == 0,1],color="red",label="Class 0")
plt.scatter(X_and[y_and == 1,0],X_and[y_and == 1,1],color="blue",label="Class 1")
plt.plot(x_values,y_values,color="black",label="Decision Boundary")
plt.xlabel("Input 1")
plt.ylabel("Input 2")
plt.title("AND Perceptron Decision Boundary")
plt.xlim(-0.5,1.5)
plt.ylim(-0.5,1.5)
plt.legend()
plt.grid()
plt.show()

print("\nPart D")
print("AND Misclassification Curve")
plt.figure(figsize=(7,5))
plt.plot(range(1,epochs + 1),and_perceptron.error_history,marker="o")
plt.xlabel("Epoch")
plt.ylabel("Misclassifications")
plt.title("AND Perceptron Convergence")
plt.grid()
plt.show()

print("\nPart E")
print("OR Truth Table")
X_or = np.array([[0,0],[0,1],[1,0],[1,1]])
y_or = np.array([0,1,1,1])
or_perceptron = Perceptron()
print("Training OR perceptron")
or_perceptron.fit(X_or,y_or,epochs,learning_rate)
or_predictions = np.array([or_perceptron.predict(x) for x in X_or])
print("\nOR Results")
print("Input1 Input2 Actual Predicted")
for i in range(len(X_or)):
    print(X_or[i,0],X_or[i,1],y_or[i],or_predictions[i])
print("Final weights:",or_perceptron.weights)
print("Final bias:",or_perceptron.bias)

print("\nPart F")
print("OR Decision Boundary")
x_values = np.linspace(-0.5,1.5,100)
if or_perceptron.weights[1] != 0:
    y_values = -(or_perceptron.weights[0] * x_values + or_perceptron.bias) / or_perceptron.weights[1]
plt.figure(figsize=(7,5))
plt.scatter(X_or[y_or == 0,0],X_or[y_or == 0,1],color="red",label="Class 0")
plt.scatter(X_or[y_or == 1,0],X_or[y_or == 1,1],color="blue",label="Class 1")
plt.plot(x_values,y_values,color="black",label="Decision Boundary")
plt.xlabel("Input 1")
plt.ylabel("Input 2")
plt.title("OR Perceptron Decision Boundary")
plt.xlim(-0.5,1.5)
plt.ylim(-0.5,1.5)
plt.legend()
plt.grid()
plt.show()

print("\nPart G")
print("OR Misclassification Curve")
plt.figure(figsize=(7,5))
plt.plot(range(1,epochs + 1),or_perceptron.error_history,marker="o")
plt.xlabel("Epoch")
plt.ylabel("Misclassifications")
plt.title("OR Perceptron Convergence")
plt.grid()
plt.show()

print("\nPart H")
print("XOR Truth Table")
X_xor = np.array([[0,0],[0,1],[1,0],[1,1]])
y_xor = np.array([0,1,1,0])
xor_perceptron = Perceptron()
print("Training XOR perceptron")
xor_perceptron.fit(X_xor,y_xor,epochs,learning_rate)
xor_predictions = np.array([xor_perceptron.predict(x) for x in X_xor])
print("\nXOR Results")
print("Input1 Input2 Actual Predicted")
for i in range(len(X_xor)):
    print(X_xor[i,0],X_xor[i,1],y_xor[i],xor_predictions[i])
print("Final weights:",xor_perceptron.weights)
print("Final bias:",xor_perceptron.bias)

print("\nPart I")
print("XOR Misclassification Curve")
plt.figure(figsize=(7,5))
plt.plot(range(1,epochs + 1),xor_perceptron.error_history,marker="o")
plt.xlabel("Epoch")
plt.ylabel("Misclassifications")
plt.title("XOR Perceptron Error Curve")
plt.grid()
plt.show()

print("\nPart J")
print("XOR Linear Separability")
print("XOR requires the following outputs:")
print("Input 00 -> Output 0")
print("Input 01 -> Output 1")
print("Input 10 -> Output 1")
print("Input 11 -> Output 0")
print("The two Class 1 points are diagonally opposite")
print("The two Class 0 points are also diagonally opposite")
print("No single straight line can separate the two classes")
print("Therefore XOR is not linearly separable")
print("A single perceptron cannot learn XOR")
plt.figure(figsize=(7,5))
plt.scatter(X_xor[y_xor == 0,0],X_xor[y_xor == 0,1],color="red",s=150,label="Class 0")
plt.scatter(X_xor[y_xor == 1,0],X_xor[y_xor == 1,1],color="blue",s=150,label="Class 1")
plt.xlabel("Input 1")
plt.ylabel("Input 2")
plt.title("XOR Linear Separability")
plt.xticks([0,1])
plt.yticks([0,1])
plt.legend()
plt.grid()
plt.show()

print("\nPart K")
print("Final Comparison")
and_final_error = and_perceptron.error_history[-1]
or_final_error = or_perceptron.error_history[-1]
xor_final_error = xor_perceptron.error_history[-1]
print("AND final misclassifications:",and_final_error)
print("OR final misclassifications:",or_final_error)
print("XOR final misclassifications:",xor_final_error)
plt.figure(figsize=(8,5))
plt.plot(range(1,epochs + 1),and_perceptron.error_history,marker="o",label="AND")
plt.plot(range(1,epochs + 1),or_perceptron.error_history,marker="s",label="OR")
plt.plot(range(1,epochs + 1),xor_perceptron.error_history,marker="^",label="XOR")
plt.xlabel("Epoch")
plt.ylabel("Misclassifications")
plt.title("Perceptron Convergence Comparison")
plt.legend()
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

print("Part A")
print("2-Layer MLP From Scratch")
print("Input Layer -> Hidden Layer -> Output Layer")
print("Hidden activation: Sigmoid")
print("Output activation: Sigmoid")
print("No deep learning framework is used")
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([[0],[1],[1],[0]])
print("Input shape:",X.shape)
print("Target shape:",y.shape)
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
def sigmoid_derivative(x):
    return x * (1 - x)
print("Sigmoid function created")
print("Sigmoid derivative function created")

print("\nPart B")
print("Forward Propagation and Backpropagation")
class MLP:
    def __init__(self,input_size,hidden_size,output_size,learning_rate):
        self.hidden_size = hidden_size
        self.learning_rate = learning_rate
        self.weights_input_hidden = np.random.randn(input_size,hidden_size) * 0.5
        self.bias_hidden = np.zeros((1,hidden_size))
        self.weights_hidden_output = np.random.randn(hidden_size,output_size) * 0.5
        self.bias_output = np.zeros((1,output_size))
        self.loss_history = []
    def forward(self,X):
        self.hidden_input = np.dot(X,self.weights_input_hidden) + self.bias_hidden
        self.hidden_output = sigmoid(self.hidden_input)
        self.output_input = np.dot(self.hidden_output,self.weights_hidden_output) + self.bias_output
        self.output = sigmoid(self.output_input)
        return self.output
    def backward(self,X,y):
        samples = len(X)
        output_error = self.output - y
        output_delta = output_error * sigmoid_derivative(self.output)
        hidden_error = np.dot(output_delta,self.weights_hidden_output.T)
        hidden_delta = hidden_error * sigmoid_derivative(self.hidden_output)
        gradient_weights_hidden_output = np.dot(self.hidden_output.T,output_delta) / samples
        gradient_bias_output = np.sum(output_delta,axis=0,keepdims=True) / samples
        gradient_weights_input_hidden = np.dot(X.T,hidden_delta) / samples
        gradient_bias_hidden = np.sum(hidden_delta,axis=0,keepdims=True) / samples
        self.weights_hidden_output -= self.learning_rate * gradient_weights_hidden_output
        self.bias_output -= self.learning_rate * gradient_bias_output
        self.weights_input_hidden -= self.learning_rate * gradient_weights_input_hidden
        self.bias_hidden -= self.learning_rate * gradient_bias_hidden
    def fit(self,X,y,epochs):
        self.loss_history = []
        for epoch in range(epochs):
            predictions = self.forward(X)
            loss = np.mean((y - predictions) ** 2)
            self.loss_history.append(loss)
            self.backward(X,y)
            if epoch == 0 or (epoch + 1) % 1000 == 0:
                print("Epoch:",epoch + 1,"Loss:",loss)
            if loss < 0.001:
                print("Converged at epoch:",epoch + 1)
                break
    def predict(self,X):
        predictions = self.forward(X)
        return np.where(predictions >= 0.5,1,0)
print("MLP class created")

print("\nPart C")
print("Training MLP on XOR")
np.random.seed(42)
hidden_neurons = 2
learning_rate = 1.0
epochs = 20000
mlp = MLP(2,hidden_neurons,1,learning_rate)
print("Hidden neurons:",hidden_neurons)
print("Learning rate:",learning_rate)
print("Maximum epochs:",epochs)
mlp.fit(X,y,epochs)
predictions = mlp.predict(X)
final_outputs = mlp.forward(X)
print("\nXOR Results")
print("Input1 Input2 Actual Output Prediction")
for i in range(len(X)):
    print(X[i,0],X[i,1],y[i,0],round(final_outputs[i,0],4),predictions[i,0])
final_loss = np.mean((y - final_outputs) ** 2)
print("Final loss:",final_loss)
print("Final weights input-hidden:")
print(mlp.weights_input_hidden)
print("Final bias hidden:")
print(mlp.bias_hidden)
print("Final weights hidden-output:")
print(mlp.weights_hidden_output)
print("Final bias output:")
print(mlp.bias_output)

print("\nPart D")
print("XOR Loss Curve")
plt.figure(figsize=(8,5))
plt.plot(range(1,len(mlp.loss_history) + 1),mlp.loss_history)
plt.xlabel("Epoch")
plt.ylabel("Mean Squared Error")
plt.title("MLP XOR Loss")
plt.grid()
plt.show()

print("\nPart E")
print("Learned XOR Decision Boundary")
x_min = -0.5
x_max = 1.5
y_min = -0.5
y_max = 1.5
x_values = np.linspace(x_min,x_max,200)
y_values = np.linspace(y_min,y_max,200)
xx,yy = np.meshgrid(x_values,y_values)
grid = np.c_[xx.ravel(),yy.ravel()]
grid_output = mlp.forward(grid)
grid_output = grid_output.reshape(xx.shape)
plt.figure(figsize=(8,6))
plt.contourf(xx,yy,grid_output,levels=[0,0.5,1],alpha=0.3)
plt.contour(xx,yy,grid_output,levels=[0.5],colors="black")
plt.scatter(X[y[:,0] == 0,0],X[y[:,0] == 0,1],color="red",s=150,label="Class 0")
plt.scatter(X[y[:,0] == 1,0],X[y[:,0] == 1,1],color="blue",s=150,label="Class 1")
plt.xlabel("Input 1")
plt.ylabel("Input 2")
plt.title("MLP Learned XOR Decision Boundary")
plt.xlim(-0.5,1.5)
plt.ylim(-0.5,1.5)
plt.legend()
plt.grid()
plt.show()

print("\nPart F")
print("Hidden Neurons Experiment")
hidden_neurons_list = [2,4,8]
learning_rate_experiment = 1.0
experiment_results = []
for hidden_neurons in hidden_neurons_list:
    np.random.seed(42)
    model = MLP(2,hidden_neurons,1,learning_rate_experiment)
    model.fit(X,y,20000)
    final_loss = model.loss_history[-1]
    epochs_needed = len(model.loss_history)
    experiment_results.append([hidden_neurons,learning_rate_experiment,epochs_needed,final_loss])
    print("Hidden neurons:",hidden_neurons)
    print("Learning rate:",learning_rate_experiment)
    print("Epochs needed:",epochs_needed)
    print("Final loss:",final_loss)

print("\nPart G")
print("Learning Rate Experiment")
hidden_neurons_experiment = 4
learning_rates = [0.01,0.1,1.0]
for learning_rate in learning_rates:
    np.random.seed(42)
    model = MLP(2,hidden_neurons_experiment,1,learning_rate)
    model.fit(X,y,100000)
    final_loss = model.loss_history[-1]
    epochs_needed = len(model.loss_history)
    experiment_results.append([hidden_neurons_experiment,learning_rate,epochs_needed,final_loss])
    print("Hidden neurons:",hidden_neurons_experiment)
    print("Learning rate:",learning_rate)
    print("Epochs needed:",epochs_needed)
    print("Final loss:",final_loss)

print("\nPart H")
print("Hidden Neurons and Learning Rate Comparison")
print("Hidden Neurons Learning Rate Epochs Final Loss")
for result in experiment_results:
    print(result[0],result[1],result[2],result[3])

print("\nPart I")
print("Complete Configuration Table")
print("Configuration Hidden Neurons Learning Rate Epochs Final Loss")
for i,result in enumerate(experiment_results):
    print(i + 1,result[0],result[1],result[2],result[3])

print("\nPart J")
print("Experiment Visualization")
configuration_names = []
epoch_values = []
for i,result in enumerate(experiment_results):
    configuration_names.append(str(result[0]) + "-" + str(result[1]))
    epoch_values.append(result[2])
plt.figure(figsize=(10,5))
plt.bar(configuration_names,epoch_values)
plt.xlabel("Hidden Neurons - Learning Rate")
plt.ylabel("Epochs to Converge")
plt.title("MLP XOR Convergence Comparison")
plt.grid(axis="y")
plt.show()

print("\nPart K")
print("Final Conclusion")
print("A single perceptron cannot solve XOR because XOR is not linearly separable")
print("The 2-layer MLP uses a hidden layer to learn a nonlinear decision boundary")
print("The sigmoid activation function provides the required nonlinearity")
print("The MLP can correctly classify the four XOR inputs")
print("Increasing hidden neurons changes the learning behavior")
print("Changing the learning rate changes the speed of convergence")
```

```python
import numpy as np
import matplotlib.pyplot as plt

print("Part A")
print("MLP Initialization Schemes")
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([[0],[1],[1],[0]])
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
def sigmoid_derivative(x):
    return x * (1 - x)
class MLP:
    def __init__(self,input_size,hidden_size,output_size,learning_rate,init_method):
        self.hidden_size = hidden_size
        self.learning_rate = learning_rate
        self.init_method = init_method
        self.initialize_weights(input_size,hidden_size,output_size)
    def initialize_weights(self,input_size,hidden_size,output_size):
        if self.init_method == "zero":
            self.weights_input_hidden = np.zeros((input_size,hidden_size))
            self.bias_hidden = np.zeros((1,hidden_size))
            self.weights_hidden_output = np.zeros((hidden_size,output_size))
            self.bias_output = np.zeros((1,output_size))
        elif self.init_method == "random":
            self.weights_input_hidden = np.random.randn(input_size,hidden_size) * 0.5
            self.bias_hidden = np.zeros((1,hidden_size))
            self.weights_hidden_output = np.random.randn(hidden_size,output_size) * 0.5
            self.bias_output = np.zeros((1,output_size))
        elif self.init_method == "xavier":
            limit = np.sqrt(6 / (input_size + hidden_size))
            self.weights_input_hidden = np.random.uniform(-limit,limit,(input_size,hidden_size))
            limit = np.sqrt(6 / (hidden_size + output_size))
            self.weights_hidden_output = np.random.uniform(-limit,limit,(hidden_size,output_size))
            self.bias_hidden = np.zeros((1,hidden_size))
            self.bias_output = np.zeros((1,output_size))
        elif self.init_method == "he":
            self.weights_input_hidden = np.random.randn(input_size,hidden_size) * np.sqrt(2 / input_size)
            self.weights_hidden_output = np.random.randn(hidden_size,output_size) * np.sqrt(2 / hidden_size)
            self.bias_hidden = np.zeros((1,hidden_size))
            self.bias_output = np.zeros((1,output_size))
    def forward(self,X):
        self.hidden_input = np.dot(X,self.weights_input_hidden) + self.bias_hidden
        self.hidden_output = sigmoid(self.hidden_input)
        self.output_input = np.dot(self.hidden_output,self.weights_hidden_output) + self.bias_output
        self.output = sigmoid(self.output_input)
        return self.output
    def backward(self,X,y):
        samples = len(X)
        output_error = self.output - y
        output_delta = output_error * sigmoid_derivative(self.output)
        hidden_error = np.dot(output_delta,self.weights_hidden_output.T)
        hidden_delta = hidden_error * sigmoid_derivative(self.hidden_output)
        gradient_weights_hidden_output = np.dot(self.hidden_output.T,output_delta) / samples
        gradient_bias_output = np.sum(output_delta,axis=0,keepdims=True) / samples
        gradient_weights_input_hidden = np.dot(X.T,hidden_delta) / samples
        gradient_bias_hidden = np.sum(hidden_delta,axis=0,keepdims=True) / samples
        self.weights_hidden_output -= self.learning_rate * gradient_weights_hidden_output
        self.bias_output -= self.learning_rate * gradient_bias_output
        self.weights_input_hidden -= self.learning_rate * gradient_weights_input_hidden
        self.bias_hidden -= self.learning_rate * gradient_bias_hidden
    def fit(self,X,y,epochs,print_epochs=False):
        self.loss_history = []
        for epoch in range(epochs):
            predictions = self.forward(X)
            loss = np.mean((y - predictions) ** 2)
            self.loss_history.append(loss)
            self.backward(X,y)
            if print_epochs and (epoch < 5):
                print("Epoch:",epoch + 1)
                print("Hidden weight matrix:")
                print(self.weights_input_hidden)
            if loss < 0.001:
                print("Converged at epoch:",epoch + 1)
                break
    def predict(self,X):
        predictions = self.forward(X)
        return np.where(predictions >= 0.5,1,0)
print("MLP class created")
print("Initialization methods: zero, random, xavier, he")

print("\nPart B")
print("Fixed Hyperparameters")
input_size = 2
hidden_size = 2
output_size = 1
learning_rate = 1.0
epochs = 20000
print("Input neurons:",input_size)
print("Hidden neurons:",hidden_size)
print("Output neurons:",output_size)
print("Learning rate:",learning_rate)
print("Maximum epochs:",epochs)

print("\nPart C")
print("Zero Initialization")
np.random.seed(42)
zero_model = MLP(input_size,hidden_size,output_size,learning_rate,"zero")
print("Initial hidden weight matrix:")
print(zero_model.weights_input_hidden)
zero_model.fit(X,y,epochs,True)
print("Final hidden weight matrix:")
print(zero_model.weights_input_hidden)
print("Final hidden biases:")
print(zero_model.bias_hidden)
print("Final output weights:")
print(zero_model.weights_hidden_output)
print("Final output bias:")
print(zero_model.bias_output)
print("Final loss:",zero_model.loss_history[-1])

print("\nPart D")
print("Zero Initialization Symmetry Breaking Failure")
print("Hidden neuron 1 weights:",zero_model.weights_input_hidden[:,0])
print("Hidden neuron 2 weights:",zero_model.weights_input_hidden[:,1])
if np.allclose(zero_model.weights_input_hidden[:,0],zero_model.weights_input_hidden[:,1]):
    print("Both hidden neurons have identical weights")
    print("The hidden neurons learn the same representation")
    print("Zero initialization fails to break symmetry")
else:
    print("Hidden neurons have different weights")

print("\nPart E")
print("Random Initialization")
np.random.seed(42)
random_model = MLP(input_size,hidden_size,output_size,learning_rate,"random")
print("Initial hidden weight matrix:")
print(random_model.weights_input_hidden)
random_model.fit(X,y,epochs)
print("Final hidden weight matrix:")
print(random_model.weights_input_hidden)
print("Final loss:",random_model.loss_history[-1])

print("\nPart F")
print("Xavier Initialization")
np.random.seed(42)
xavier_model = MLP(input_size,hidden_size,output_size,learning_rate,"xavier")
print("Initial hidden weight matrix:")
print(xavier_model.weights_input_hidden)
xavier_model.fit(X,y,epochs)
print("Final hidden weight matrix:")
print(xavier_model.weights_input_hidden)
print("Final loss:",xavier_model.loss_history[-1])

print("\nPart G")
print("He Initialization")
np.random.seed(42)
he_model = MLP(input_size,hidden_size,output_size,learning_rate,"he")
print("Initial hidden weight matrix:")
print(he_model.weights_input_hidden)
he_model.fit(X,y,epochs)
print("Final hidden weight matrix:")
print(he_model.weights_input_hidden)
print("Final loss:",he_model.loss_history[-1])

print("\nPart H")
print("XOR Predictions")
models = [
    ["Zero",zero_model],
    ["Random",random_model],
    ["Xavier",xavier_model],
    ["He",he_model]
]
print("Initialization Input1 Input2 Actual Prediction")
for name,model in models:
    predictions = model.predict(X)
    print("\n",name)
    for i in range(len(X)):
        print(X[i,0],X[i,1],y[i,0],predictions[i,0])

print("\nPart I")
print("Loss Curve Comparison")
plt.figure(figsize=(10,6))
plt.plot(zero_model.loss_history,label="Zero")
plt.plot(random_model.loss_history,label="Random")
plt.plot(xavier_model.loss_history,label="Xavier")
plt.plot(he_model.loss_history,label="He")
plt.xlabel("Epoch")
plt.ylabel("Mean Squared Error")
plt.title("Loss Comparison for Initialization Methods")
plt.legend()
plt.grid()
plt.show()

print("\nPart J")
print("Final Loss Comparison")
initialization_names = ["Zero","Random","Xavier","He"]
final_losses = [
    zero_model.loss_history[-1],
    random_model.loss_history[-1],
    xavier_model.loss_history[-1],
    he_model.loss_history[-1]
]
print("Initialization Final Loss")
for i in range(len(initialization_names)):
    print(initialization_names[i],final_losses[i])
plt.figure(figsize=(8,5))
plt.bar(initialization_names,final_losses)
plt.xlabel("Initialization Method")
plt.ylabel("Final Loss")
plt.title("Final Loss Comparison")
plt.grid(axis="y")
plt.show()

print("\nPart K")
print("Epochs to Convergence")
epochs_to_convergence = []
for name,model in models:
    convergence_epoch = next((i + 1 for i,loss in enumerate(model.loss_history) if loss < 0.001),epochs)
    epochs_to_convergence.append(convergence_epoch)
    print(name,"epochs:",convergence_epoch)

print("\nPart L")
print("Convergence Table")
print("Initialization Epochs to Convergence Final Loss")
for i in range(len(initialization_names)):
    print(initialization_names[i],epochs_to_convergence[i],final_losses[i])
plt.figure(figsize=(8,5))
plt.bar(initialization_names,epochs_to_convergence)
plt.xlabel("Initialization Method")
plt.ylabel("Epochs to Convergence")
plt.title("Epochs to Convergence Comparison")
plt.grid(axis="y")
plt.show()

print("\nPart M")
print("Final Conclusion")
print("Zero initialization gives identical hidden neuron weights")
print("Identical hidden neurons cause symmetry and prevent useful feature learning")
print("Random initialization breaks the symmetry between hidden neurons")
print("Xavier initialization scales weights according to the input and output sizes")
print("He initialization uses a larger variance suitable for ReLU networks")
print("The same XOR problem can behave differently under different initialization schemes")
print("The loss curves and convergence results show the effect of initialization")
```
