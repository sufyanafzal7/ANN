Task 6: Conditional Statements

```python
print("Part A")
marks = 85

if marks >= 80:
    print("Grade: A")
elif marks >= 70:
    print("Grade: B")
elif marks >= 60:
    print("Grade: C")
elif marks >= 50:
    print("Grade: D")
else:
    print("Grade: F")

print("\nPart B")
number = -5

if number > 0:
    print("The number is positive.")
elif number < 0:
    print("The number is negative.")
else:
    print("The number is zero.")

print("\nPart C")
year = 2024

if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0:
            print("The year is a leap year.")
        else:
            print("The year is not a leap year.")
    else:
        print("The year is a leap year.")
else:
    print("The year is not a leap year.")

print("\nPart D")
cgpa = 3.7
attendance = 85

if cgpa > 3.5 and attendance > 80:
    print("Student is eligible for the scholarship.")
elif cgpa > 3.5 or attendance > 80:
    print("Student partially meets the scholarship criteria.")
else:
    print("Student is not eligible for the scholarship.")
```

Task 7: Loops with Break and Continue

```python
print("Part A")
for number in range(1, 31):
    if number % 4 == 0:
        continue
    print(number)

print("\nPart B")
for number in range(1, 51):
    if number > 20:
        break
    print(number)

print("\nPart C")
number = 7
counter = 1

while counter <= 10:
    print(number, "x", counter, "=", number * counter)
    counter += 1

print("\nPart D")
total = 0

for number in range(1, 101):
    if number % 2 == 0:
        total += number

print("Sum of even numbers:", total)
```

Task 8: Functions – Basics

```python
print("Part A")
def calculate_area(length, width=1):
    return length * width

print("Area:", calculate_area(5, 4))

print("\nPart B")
print("Area:", calculate_area(10))
print("Area:", calculate_area(6, 3))

area = calculate_area(8, 5)
print("Stored area:", area)

print("\nPart C")
def is_even(number):
    if number % 2 == 0:
        return True
    else:
        return False

print("Is 10 even?", is_even(10))
print("Is 7 even?", is_even(7))

print("\nPart D")
def add_item(items):
    print("Inside function before adding:", items)
    items.append("NLP")
    print("Inside function after adding:", items)

subjects = ["Python", "AI", "Database"]
print("Before function call:", subjects)
add_item(subjects)
print("After function call:", subjects)
```

Task 9: Classes and Inheritance

```python
print("Part A")
class Course:
    def __init__(self, course_name, credit_hours):
        self.course_name = course_name
        self.credit_hours = credit_hours

    def display_info(self):
        print("Course Name:", self.course_name)
        print("Credit Hours:", self.credit_hours)

print("\nPart B")
course1 = Course("Artificial Intelligence", 3)
course2 = Course("Python Programming", 3)

course1.display_info()
course2.display_info()

print("\nPart C")
class LabCourse(Course):
    def __init__(self, course_name, credit_hours, lab_hours):
        super().__init__(course_name, credit_hours)
        self.lab_hours = lab_hours

print("\nPart D")
class LabCourse(Course):
    def __init__(self, course_name, credit_hours, lab_hours):
        super().__init__(course_name, credit_hours)
        self.lab_hours = lab_hours

    def display_info(self):
        print("Course Name:", self.course_name)
        print("Credit Hours:", self.credit_hours)
        print("Lab Hours:", self.lab_hours)

lab_course = LabCourse("Natural Language Processing", 3, 2)
lab_course.display_info()
```
