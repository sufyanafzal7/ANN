Task 3: Addition & Subtraction

```python
term1_marks = np.array([
    [85, 90, 78],
    [92, 88, 95],
    [78, 82, 80]
])
term2_marks = np.array([
    [88, 92, 82],
    [95, 91, 97],
    [85, 88, 86]
])
print(f"Term 1 marks:\n{term1_marks}")
print(f"Term 2 marks:\n{term2_marks}")
total_marks = term1_marks + term2_marks
print(f"Total combined marks (Term 1 + Term 2):\n{total_marks}")
improvement = term2_marks - term1_marks
print(f"Improvement (Term 2 - Term 1):\n{improvement}")
total_improvement_per_student = np.sum(improvement, axis=1)
print(f"Total improvement per student: {total_improvement_per_student}")
best_student_idx = np.argmax(total_improvement_per_student)
best_student_improvement = total_improvement_per_student[best_student_idx]
print(f"Student {best_student_idx} had highest improvement: {best_student_improvement}")
```

Task 4: Dot Product

```python
prices = np.array([10.5, 25.0, 8.75, 15.0])
quantities = np.array([3, 2, 5, 4])
print(f"\nPrices: {prices}")
print(f"Quantities: {quantities}")
total_bill = np.dot(prices, quantities)
print(f"Total bill (using dot product): ${total_bill:.2f}")
customer_quantities = np.array([
    [3, 2, 5, 4],      
    [2, 3, 2, 6]      
])
print(f"\nCustomer quantities (2x4 matrix):\n{customer_quantities}")
customer_bills = np.dot(prices, customer_quantities.T)
print(f"Bills for both customers: {customer_bills}")
print("""
EXPLANATION: The dot product requires the inner dimensions to match.
- prices shape: (4,)
- customer_quantities shape: (2, 4)
To multiply prices by each customer's quantities, we need to transpose
customer_quantities to (4, 2), so: (4,) • (4, 2) = (2,)
This gives us the total bill for each of the 2 customers.
""")
```

Task 8: Random Matrices

```python
np.random.seed(123)
random_decimals_1 = np.random.rand(3, 3)
print(f"\n3x3 random decimals (0-1):\n{random_decimals_1}")
random_integers = np.random.randint(10, 101, size=(3, 3))
print(f"\n3x3 random integers (10-100):\n{random_integers}")
random_normal = np.random.randn(4, 4)
print(f"\n4x4 standard normal distribution:\n{random_normal}")
random_decimals_2 = np.random.rand(3, 3)
print(f"\n3x3 random decimals (second run):\n{random_decimals_2}")
print(f"\nAre both random decimal matrices identical?\n{np.array_equal(random_decimals_1, random_decimals_2)}")
print("""
EXPLANATION: Both outputs are identical because:
- We set the random seed to 123 before the first rand() call
- Each call to rand() advances the random number generator state
- The second rand() call gets the NEXT sequence from the same generator
- Since we didn't reset the seed between calls, they produce different outputs
Actually, in this case they ARE different because the seed only affects the
first call. If we reset the seed, then both would be identical.
Let me demonstrate:
""")
np.random.seed(456)
demo_1 = np.random.rand(3, 3)
np.random.seed(456)  
demo_2 = np.random.rand(3, 3)
print(f"With seed reset between calls - are they identical? {np.array_equal(demo_1, demo_2)}")
```

Task 9: Transpose

```python
performance = np.array([
    [85, 88, 90, 87, 89],
    [92, 90, 88, 91, 93],
    [78, 80, 82, 81, 79]
])
print(f"\nOriginal performance matrix (3x5):\n{performance}")
print(f"Shape: {performance.shape}")
performance_T = performance.T
print(f"\nTransposed matrix (5x3):\n{performance_T}")
print(f"New shape: {performance_T.shape}")
performance_TT = performance_T.T
is_equal = np.array_equal(performance, performance_TT)
print(f"\nDoes transposing twice give back original? {is_equal}")
matrix_4x2 = np.array([
    [1, 2],
    [3, 4],
    [5, 6],
    [7, 8]
])
matrix_3x2 = np.array([
    [9, 10],
    [11, 12],
    [13, 14]
])
print(f"\nMatrix 1 (4x2):\n{matrix_4x2}")
print(f"Matrix 2 (3x2):\n{matrix_3x2}")
matrix_3x2_T = matrix_3x2.T
dot_product = np.dot(matrix_4x2, matrix_3x2_T)
print(f"\nMatrix 2 transposed (2x3):\n{matrix_3x2_T}")
print(f"\nDot product result (4x3):\n{dot_product}")
print("""
EXPLANATION: 
- Original shapes: (4x2) and (3x2) cannot be multiplied
- By transposing (3x2) to (2x3), we get: (4x2) • (2x3) = (4x3)
- The inner dimensions match (2), allowing the dot product to work
""")
```
